# Tokenizers testing
- it is example of workflow with my self written tokkenizers

```python
from src import BasicTokenizer, RegexTokenizer, GPT4Tokenizer
import tiktoken
```

# Load the text data

```python
file_path = "./dataset/taylorswift.txt"
file = open(file=file_path, mode="r")
text = file.read()
text = text + "<special_token1><special_token2>  <special_token3><special_token4><|endoftext|>"
print(text[:50])
```

# Basic Tokenizer

```python
tokenizer_basic = BasicTokenizer()

basic_vocab_size = 256

tokenizer_basic.train(text=text, vocab_size=basic_vocab_size, verbose=True)
```

```python
text_encoded = tokenizer_basic.encode(text=text)
text_reversed = tokenizer_basic.decode(text_encoded)
print(text_encoded[-50:])
```

```python
if text_reversed == text:
    print("[PASS] Text == Text_after_tokenization")
else:
    print("[FAILED] Text after tokenization steps are not equal")
```

# Regex Tokenizer (with special token)

```python
tokenizer_regex = RegexTokenizer()

regex_vocab_size = 256

tokenizer_regex.register_special_token({"<special_token1>": regex_vocab_size+1, "<special_token2>": regex_vocab_size+2, "<special_token3>": regex_vocab_size+3, "<special_token4>": regex_vocab_size+4})
tokenizer_regex.train(text=text, vocab_size=256,verbose=True)
```

```python
text_encoded = tokenizer_regex.encode(text=text, allow_special='all')
text_reversed = tokenizer_regex.decode(text_encoded)
print(text_encoded[-50:])
```

```python
if text_reversed == text:
    print("[PASS] Text == Text_after_tokenization")
else:
    print("[FAILED] Text after tokenization steps are not equal")
```

# GPT4 tokenizer

```python
tokenizer_gpt = GPT4Tokenizer()
tiktoken_gpt = tiktoken.get_encoding("cl100k_base")
```

```python
my_tokens = tokenizer_gpt.encode(text, allow_special='all')
print(my_tokens[-20:])
my_tokens_reversed = tokenizer_gpt.decode(my_tokens)

if my_tokens_reversed == text:
    print("[PASS] Text == Text_after_tokenization")
    print(my_tokens_reversed[-100:])
else:
    print("[FAILED] Text after tokenization steps are not equal")
    print(my_tokens_reversed[-100:])
```

```python
tiktoken_tokens = tiktoken_gpt.encode(text, allowed_special='all')
tiktoken_tokens_reversed = tiktoken_gpt.decode(tiktoken_tokens)

if tiktoken_tokens == my_tokens:
    print("[PASS] Tiktoken tokens == My Tokens")
    print(f"Tiktoken: {tiktoken_tokens_reversed[-10:]}\nMy Tokens: {my_tokens_reversed[-10:]}")
else:
    print("[FAILED] Text after tokenization steps are not equal")
    print(f"Tiktoken: {tiktoken_tokens_reversed[-10:]}\nMy Tokens: {my_tokens_reversed[-10]}")

if my_tokens_reversed == tiktoken_tokens_reversed:
    print("[PASS] Tiktoken reversed tokens == My reversed Tokens")
    print(f"Tiktoken: {tiktoken_tokens_reversed[-10:]}\nMy Tokens: {my_tokens_reversed[-10:]}")
else:
    print("[FAILED] Text after tokenization steps are not equal")
    print(f"Tiktoken: {tiktoken_tokens_reversed[-10:]}\nMy Tokens: {my_tokens_reversed[-10]}")
```
